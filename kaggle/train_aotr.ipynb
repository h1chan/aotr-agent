# AoTR agent: обучение на Kaggle GPU

Перед запуском:
1. Загрузи код проекта как Kaggle Dataset (zip с папкой `aotr-agent`).
2. Загрузи датасеты (`ds_e1`, `ds_e2`, `ds_dagger`) как второй Dataset.
3. Подключи оба датасета к ноутбуку справа.
4. В настройках ноутбука выбери Accelerator: GPU T4 x2.

Ниже поправь пути под свои имена датасетов.

```python
!nvidia-smi
import torch
print('torch', torch.__version__, 'cuda', torch.cuda.is_available())
```

```python
import os, glob, shutil, zipfile

CODE_ZIP = '/kaggle/input/aotr-agent/aotr-agent.zip'   # поправь
DATA_DIR = '/kaggle/input/aotr-datasets'               # поправь

if os.path.exists('/kaggle/working/aotr-agent'):
    shutil.rmtree('/kaggle/working/aotr-agent')
with zipfile.ZipFile(CODE_ZIP) as z:
    z.extractall('/kaggle/working')

print('код:', os.listdir('/kaggle/working/aotr-agent')[:10])
print('данные:', os.listdir(DATA_DIR))
```

```python
%%bash
cd /kaggle/working/aotr-agent
python -u train.py \
  /kaggle/input/aotr-datasets/ds_e1 \
  /kaggle/input/aotr-datasets/ds_e2 \
  /kaggle/input/aotr-datasets/ds_dagger \
  --epochs 12 --batch 128 --device cuda \
  --out /kaggle/working/policy_v3.pt
```
